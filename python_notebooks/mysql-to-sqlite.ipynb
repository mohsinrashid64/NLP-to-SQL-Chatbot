```python
# Install: pip install mysql-connector-python
import mysql.connector
import sqlite3

# Connect to MySQL
mysql_conn = mysql.connector.connect(
    host= 'zkw-test-mysql.mysql.database.azure.com',
    port= 3306,
    user= 'mysqladmin',
    password= 'ZkwTest2026Pass1',
    database= 'zkwmbdb_global_elunic'
)

# Create SQLite database
sqlite_conn = sqlite3.connect("zkw_test.db")

```

```python

# Get all tables
mysql_cursor = mysql_conn.cursor()
mysql_cursor.execute("SHOW TABLES")

```

```python
tables = mysql_cursor.fetchall()
for (table_name,) in tables:
    # Get table data
    mysql_cursor.execute(f"SELECT * FROM `{table_name}`")
    rows = mysql_cursor.fetchall()
    
    # Get column names
    columns = [desc[0] for desc in mysql_cursor.description]
    
    # Create table in SQLite (simplified - adjust types as needed)
    cols = ", ".join([f'"{col}" TEXT' for col in columns])
    sqlite_conn.execute(f'CREATE TABLE IF NOT EXISTS "{table_name}" ({cols})')
    
    # Insert data
    placeholders = ", ".join(["?" for _ in columns])
    sqlite_conn.executemany(f'INSERT INTO "{table_name}" VALUES ({placeholders})', rows)

sqlite_conn.commit()
print("Exported to zkw_test.db")
```
